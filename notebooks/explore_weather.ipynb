```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv("C:/Users/IDI/Documents/integrated-env-energy-health-pipeline/data/helsinki_weather_april_2025.csv")

# Convert datetime column
df["datetime"] = pd.to_datetime(df["datetime"])
```

```python
plt.figure(figsize=(14, 6))

# Temperature line
plt.plot(df["datetime"], df["temp"], label="Temperature (°C)", color="tab:blue")

# Humidity line
plt.plot(df["datetime"], df["humidity"], label="Humidity (%)", color="tab:orange")

plt.title("Temperature and Humidity in Helsinki (April 2025)")
plt.xlabel("Date")
plt.ylabel("Value")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
fig, ax1 = plt.subplots(figsize=(14, 6))

# Temperature on left Y-axis
ax1.plot(df["datetime"], df["temp"], color="tab:blue", label="Temperature (°C)")
ax1.set_xlabel("Date")
ax1.set_ylabel("Temperature (°C)", color="tab:blue")
ax1.tick_params(axis="y", labelcolor="tab:blue")

# Create second Y-axis for humidity
ax2 = ax1.twinx()
ax2.plot(df["datetime"], df["humidity"], color="tab:orange", label="Humidity (%)")
ax2.set_ylabel("Humidity (%)", color="tab:orange")
ax2.tick_params(axis="y", labelcolor="tab:orange")

# Title and layout
plt.title("Temperature and Humidity in Helsinki (Dual Y-Axis, April 2025)")
fig.tight_layout()
plt.show()
```

```python
# Ensure datetime is in datetime format
df["datetime"] = pd.to_datetime(df["datetime"])

# Extract just the date (without time)
df["date"] = df["datetime"].dt.date

# Group by date and compute average temperature
daily_avg_temp = df.groupby("date")["temp"].mean().reset_index()

# Plot
plt.figure(figsize=(14, 6))
plt.plot(daily_avg_temp["date"], daily_avg_temp["temp"], marker="o", linestyle="-", color="tab:blue")
plt.title("Daily Average Temperature in Helsinki (April 2025)")
plt.xlabel("Date")
plt.ylabel("Average Temperature (°C)")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Group by date and sum precipitation
daily_precip = df.groupby("date")["precip"].sum().reset_index()

# Plot
plt.figure(figsize=(14, 6))
plt.bar(daily_precip["date"], daily_precip["precip"], color="tab:cyan")
plt.title("Total Daily Precipitation in Helsinki (April 2025)")
plt.xlabel("Date")
plt.ylabel("Precipitation (mm)")
plt.xticks(rotation=45)
plt.grid(True, axis="y")
plt.tight_layout()
plt.show()
```

```python
# Plot humidity trend
plt.figure(figsize=(12, 5))
plt.plot(df["datetime"], df["humidity"], label="Humidity (%)", color="tab:green")

plt.title("Hourly Humidity in Helsinki (April 2025)")
plt.xlabel("Date and Time")
plt.ylabel("Humidity (%)")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Load the dataset again if needed
df = pd.read_csv("C:/Users/IDI/Documents/integrated-env-energy-health-pipeline/data/helsinki_weather_april_2025.csv")
df["datetime"] = pd.to_datetime(df["datetime"])
```

```python
plt.figure(figsize=(12, 5))
plt.plot(df["datetime"], df["humidity"], label="Humidity (%)", color="tab:green")
plt.plot(df["datetime"], df["dew"], label="Dew Point (°C)", color="tab:orange")

plt.title("Humidity and Dew Point in Helsinki (April 2025)")
plt.xlabel("Date")
plt.ylabel("Value")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))
plt.plot(df["datetime"], df["windspeed"], label="Wind Speed (km/h)", color="tab:purple")

plt.title("Wind Speed in Helsinki (April 2025)")
plt.xlabel("Date")
plt.ylabel("Wind Speed (km/h)")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 6))

plt.bar(daily_avg["date"], daily_avg["precip"], color="tab:purple")

plt.title("Daily Precipitation in Helsinki – April 2025")
plt.xlabel("Date")
plt.ylabel("Total Precipitation (mm)")
plt.xticks(rotation=45)
plt.grid(True, axis="y")
plt.tight_layout()
plt.show()
```

```python
# Create a new column with just the date part
df["date"] = df["datetime"].dt.date

# Group by date and compute daily averages
daily_summary = df.groupby("date").agg({
    "temp": "mean",
    "humidity": "mean",
    "dew": "mean",
    "precip": "sum"
}).reset_index()

# Rename columns for clarity
daily_summary.columns = ["Date", "Avg Temp (°C)", "Avg Humidity (%)", "Avg Dew (°C)", "Total Precip (mm)"]

# Show the result
print("✅ Daily summary created")
daily_summary.head()
```

```python
# Plot daily average temperature, humidity, and total precipitation
plt.figure(figsize=(14, 6))

# Plot average temperature
plt.plot(daily_summary["Date"], daily_summary["Avg Temp (°C)"], label="Avg Temp (°C)", color="tab:red", marker="o")

# Plot average humidity
plt.plot(daily_summary["Date"], daily_summary["Avg Humidity (%)"], label="Avg Humidity (%)", color="tab:blue", marker="x")

# Plot total daily precipitation on a second y-axis
plt.bar(daily_summary["Date"], daily_summary["Total Precip (mm)"], label="Total Precip (mm)", color="tab:green", alpha=0.4)

plt.title("Daily Weather Summary for Helsinki (April 2025)")
plt.xlabel("Date")
plt.ylabel("Temperature / Humidity")
plt.xticks(rotation=45)
plt.legend(loc="upper left")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Define a threshold for rainfall
rain_threshold = 1.0  # millimeter

# Filter rainy days
rainy_days = daily_summary[daily_summary["Total Precip (mm)"] >= rain_threshold]

# Filter dry days
dry_days = daily_summary[daily_summary["Total Precip (mm)"] < rain_threshold]

# Display the number of rainy and dry days
print(f"🌧️ Rainy days: {len(rainy_days)}")
print(f"☀️ Dry days: {len(dry_days)}")

# Show a few examples of each
print("\nRainy Days:")
display(rainy_days.head())

print("\nDry Days:")
display(dry_days.head())
```

```python
plt.figure(figsize=(12, 5))
plt.bar(daily_summary["Date"], daily_summary["Total Precip (mm)"], color="tab:blue")

plt.title("🌧️ Daily Rainfall in Helsinki (April 2025)")
plt.xlabel("Date")
plt.ylabel("Total Precipitation (mm)")
plt.xticks(rotation=45)
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Find hottest and coldest hourly records
hottest = df.loc[df["temp"].idxmax()]
coldest = df.loc[df["temp"].idxmin()]

# Display results
print("🔥 Hottest recorded hour:")
print(hottest[["datetime", "temp", "feelslike", "conditions"]])
print("\n❄️ Coldest recorded hour:")
print(coldest[["datetime", "temp", "feelslike", "conditions"]])
```

```python
# Locate hottest and coldest records
hottest = df.loc[df["temp"].idxmax()]
coldest = df.loc[df["temp"].idxmin()]

# Create the temperature trend plot
plt.figure(figsize=(12, 5))
plt.plot(df["datetime"], df["temp"], label="Temperature (°C)", color="tab:blue")

# Mark the hottest hour
plt.scatter(hottest["datetime"], hottest["temp"], color="red", label=f"Hottest: {hottest['temp']}°C", zorder=5)
plt.text(hottest["datetime"], hottest["temp"] + 0.5, "🔥", fontsize=14)

# Mark the coldest hour
plt.scatter(coldest["datetime"], coldest["temp"], color="cyan", label=f"Coldest: {coldest['temp']}°C", zorder=5)
plt.text(coldest["datetime"], coldest["temp"] + 0.5, "❄️", fontsize=14)

# Final chart setup
plt.title("Temperature in Helsinki - Hottest and Coldest Hours (April 2025)")
plt.xlabel("Date")
plt.ylabel("Temperature (°C)")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
